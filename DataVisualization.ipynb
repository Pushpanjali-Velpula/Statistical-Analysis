```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
price=[10,30,5,5,25]
sales=[5000,3500,2000,4500,2500]
print("Mean:",np.mean(price))
print("Median:",np.median(price))
print("Mode:",pd.Series(price).mode())
print("Variance:",np.var(price))
print("Standard Deviation:",np.std(price))
x=plt.title("Bar chart")
plt.xlabel("Price")
plt.ylabel("Sales")
plt.bar(price,sales)
plt.show()
x=plt.title("Scatter plot")
plt.xlabel("Price")
plt.ylabel("Sales")
plt.scatter(price,sales)
plt.show()
x=plt.title("Pie chart")
plt.pie(price,labels=price)
plt.show()
```
