```python
def even_odd(num):
  if num%2==0:
    print("even")
  else:
    print("odd")
```

```python
even_odd(26)
```

```python
even_odd(99)
```

```python
def fact(n):
  f=1
  for i in range(1,n+1):
    f=f*i
  print(f)
```

```python
fact(5)
```

```python
fact(10)
```

```python
def naturalSum(n):
  num=0
  for i in range(1,n+1):
    num=num+i
  return num
```

```python
naturalSum(10)
```
